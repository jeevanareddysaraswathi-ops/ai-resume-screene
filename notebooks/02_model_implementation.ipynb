```python
# Step 3 — TF-IDF Vectorization and Cosine Similarity
resume_folder = "../backend/uploads"
resume_files = [f for f in os.listdir(resume_folder) if f.endswith('.docx') or f.endswith('.pdf')]

jd_text = jd_processed["processed_text"]

results = []
for filename in resume_files:
    file_path = f"{resume_folder}/{filename}"
    raw_text = extract_text(file_path)
    resume_processed = preprocess(raw_text)
    resume_text = resume_processed["processed_text"]

    vectorizer = TfidfVectorizer()
    vectors = vectorizer.fit_transform([jd_text, resume_text])
    score = cosine_similarity(vectors[0], vectors[1])[0][0]

    results.append({
        "Candidate": filename.replace("_Resume.docx", "").replace("_", " "),
        "Score (%)": round(float(score) * 100, 2),
        "Matched Skills": len(set(jd_processed["skills"]) & set(resume_processed["skills"]))
    })

df = pd.DataFrame(results).sort_values("Score (%)", ascending=False)
print("🏆 CANDIDATE RANKINGS:")
print("=" * 50)
print(df.to_string(index=False))
```
